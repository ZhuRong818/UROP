# 1. Control Flow

A program normally runs from top to bottom.

**Control flow** lets the program decide:

- whether a block of code should run;
- which branch should run;
- how many times a block should repeat.

A useful mental model:

```text
Input
  ↓
Check a condition
 /           \
True         False
 ↓             ↓
Action A      Action B
```

For business problems, this often looks like:

> If spending is at least $100, give free delivery.

or

> For every transaction, add its amount to total revenue.

# 3. Comparison Operators

Comparison operators compare two values and return `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |

A very important beginner distinction:

```python
price = 20      # assignment
price == 20     # comparison
```

`=` stores a value.

`==` asks whether two values are equal.


````python
revenue = 12000

print(revenue > 10000)
print(revenue == 10000)
print(revenue != 5000)
print(revenue <= 15000)
````

# 4. Membership Operators: `in` and `not in`

Use:

- `in` to check whether a value exists inside a sequence;
- `not in` to check whether it does not exist.

Example:


````python
market = "Singapore"
target_markets = ["Singapore", "Malaysia", "Thailand"]

print(market in target_markets)
print("Japan" not in target_markets)
````

Strings are also sequences.

Membership tests on strings are **case-sensitive**.


````python
sentence = "All work and no play makes Jack a dull boy."

print("work" in sentence)
print("Work" in sentence)
print("John" not in sentence)
````

# 5. Logical Operators: `and`, `or`, `not`

Logical operators combine Boolean expressions.

### `and`
Returns `True` only if **both** conditions are true.

### `or`
Returns `True` if **at least one** condition is true.

### `not`
Reverses the Boolean value.


````python
spending = 1200
active = True

premium_customer = spending >= 1000 and active
print(premium_customer)

is_member = False
free_delivery = is_member or spending >= 50
print(free_delivery)

is_suspended = False
print(not is_suspended)
````

A quick memory rule:

```text
AND → all required
OR  → at least one required
NOT → opposite
```

# 6. The `if` Statement

Use `if` when some code should run **only when a condition is true**.

General form:

```python
if condition:
    code_to_run
```

Important details:

- `if` is followed by a Boolean condition.
- The line ends with `:`.
- The code inside the `if` block is indented.


````python
price = 35

if price >= 30:
    print("Eligible for discount")
    price = price * (1 - 0.15)

print("Final price:", price)
````

## Indentation matters

In Python, indentation is part of the program logic.

```python
if condition:
    this_runs_only_if_true
    this_also_runs_only_if_true

this_runs_after_the_if
```

# 8. The `if-elif-else` Statement

Use `elif` when there are multiple possible branches.

Python checks conditions from top to bottom and executes the **first matching branch**.

```python
if condition1:
    ...
elif condition2:
    ...
elif condition3:
    ...
else:
    ...
```


````python
spending = 2500

if spending >= 5000:
    segment = "Platinum"
elif spending >= 2000:
    segment = "Gold"
elif spending >= 500:
    segment = "Silver"
else:
    segment = "Standard"

print(segment)
````

# 9. Nested Conditional Statements

A conditional statement can appear inside another conditional statement.

This is useful when one decision depends on another.

Example:

- first decide whether the customer is a member;
- then use spending to decide the discount.


````python
is_member = True
spending = 120

if is_member:
    if spending >= 100:
        discount = 0.20
    else:
        discount = 0.10
else:
    if spending >= 100:
        discount = 0.05
    else:
        discount = 0

print("Discount rate:", discount)
````

More indentation means a deeper level of logic.

```text
outer decision
    inner decision
        inner action
```

# 10. `for` Loops

A `for` loop repeats code for each item in a sequence.

General form:

```python
for item in sequence:
    repeated_code
```

One trip through the loop is called an **iteration**.


## Iterating over a string


````python
word = "data"

for char in word:
    print(char)
````

The loop behaves conceptually like this:

```text
1st iteration → char = "d"
2nd iteration → char = "a"
3rd iteration → char = "t"
4th iteration → char = "a"
```


## Business example: process transactions


````python
transactions = [120, 85, 210, 160]

for amount in transactions:
    print("Processing transaction:", amount)
````

# 11. Accumulating Values in a Loop

A very common pattern is to build a running total.

Start with:

```python
total = 0
```

Then repeatedly add each value.


````python
sales = [120, 85, 210, 160]
total = 0

for amount in sales:
    total += amount

print("Total sales:", total)
````

The variable `total` is sometimes called an **accumulator**.

# 12. The `range()` Function

`range()` generates a sequence of integers.

The main forms are:

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

Important rule:

> The `stop` value is excluded.


## `range(stop)`


````python
for i in range(5):
    print(i)
````

This produces:

```text
0, 1, 2, 3, 4
```

not `0` through `5`.


## `range(start, stop)`


````python
for i in range(1, 5):
    print(i)
````

This produces:

```text
1, 2, 3, 4
```


## `range(start, stop, step)`


````python
for i in range(1, 9, 2):
    print(i)
````

This produces:

```text
1, 3, 5, 7
```

A useful memory rule:

```text
range(start, stop, step)

start → where we begin
stop  → excluded
step  → how much we move each time
```

# 13. `continue`

`continue` means:

> Skip the rest of the **current iteration** and move to the next iteration.

Example: ignore zero-value transactions.


````python
transactions = [100, 0, 250, 0, 80]

for amount in transactions:
    if amount == 0:
        continue

    print("Processing:", amount)
````

Another lesson-style idea is skipping spaces in a person's name:


````python
name = "Steven Curry"

for letter in name:
    if letter == " ":
        continue
    print(letter)
````

# 14. `break`

`break` means:

> Stop the **entire loop immediately**.

Compare:

```text
continue → skip one iteration
break    → stop the whole loop
```


````python
sales = [100, 250, 400, 1200, 300]

for amount in sales:
    if amount >= 1000:
        print("Large transaction found:", amount)
        break
````

# 15. Combining Loops and Conditions

Real programs often combine:

- a loop to process many values;
- an `if` statement to decide what to do with each one.


````python
transactions = [120, -30, 0, 250, 80]

valid_total = 0

for amount in transactions:
    if amount <= 0:
        continue
    valid_total += amount

print("Valid transaction total:", valid_total)
````

# 16. Lesson-style Repeated Sum Example

The lesson uses repeated addition of:

```text
1 + 1/2 + 1/3 + ...
```

to motivate loops.

Instead of writing:


````python
kills = 0
kills += 1/1
kills += 1/2
kills += 1/3

print(kills)
````

we can write:


````python
kills = 0

for i in range(1, 4):
    kills += 1/i

print(kills)
````

And if we want to stop once a threshold is reached:


````python
n = 50
kills = 0

for i in range(1, n):
    kills += 1/i
    if kills >= 4:
        break

print("Minimum number of players:", i + 1)
````
